```python
import PyPDF2

# Open the PDF file
with open('Information-Retrieval-in-High-Dimensional-Data/Lab_1.pdf', 'rb') as file:
    reader = PyPDF2.PdfReader(file)
    # Print the number of pages
    print(len(reader.pages))
    # Extract text from the first page
    page = reader.pages[0]
    print(page.extract_text())
```
